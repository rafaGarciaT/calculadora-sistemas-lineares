No projeto, utilizamos ndarrays para representar os sistemas lineares como matrizes aumentadas. Elas são escalonadas e resolvidas usando as funções que implementamos.

Vamos ver como isso funciona na prática com alguns exemplos demonstrativos.

## 3.1 Exemplo SPD
Sistema original:
$$ \huge \begin{cases} x + y + z = 6 \\ 2x - y + z = 3 \\ x + 2y - z = 3 \end{cases} $$

Representação matricial usando ndarrays:
$$ \large
\begin{pmatrix} 1 & 1 & 1 \\ 2 & -1 & 1 \\ 1 & 2 & -1 \end{pmatrix}
\begin{pmatrix} x \\ y \\ z \end{pmatrix}
=
\begin{pmatrix} 6 \\ 3 \\ 3 \end{pmatrix}
\rightarrow
\begin{pmatrix} 1 & 1 & 1 & 6 \\ 2 & -1 & 1 & 3 \\ 1 & 2 & -1 & 3 \end{pmatrix}
$$

Setup e resolução:

````python
spd_sistema = np.array([[1, 1, 1, 6], [2, -1, 1, 3], [1, 2, -1, 3]], dtype=float)
print(f"==========Sistema original==========")
imprimir_sistema(spd_sistema)
spd_sistema_escalonado = escalonar_sistema(spd_sistema)
print(f"\n\n==========Sistema escalonado==========")
imprimir_sistema(spd_sistema_escalonado)
spd_resultado = resolver(spd_sistema_escalonado)
print(f"\n\nResultado:")
imprimir_resultado(spd_resultado)
````

Resultado final:
$$ \huge
\begin{align}
x=\frac{9}{7}\ \text{ou}\ x\approx 1{,}2857
&&
y=\frac{15}{7}\ \text{ou}\ y\approx 2{,}1429
&&
z=\frac{18}{7}\ \text{ou}\ z\approx 2{,}5714
\end{align}
$$

## 3.2 Exemplo SPI
Sistema original:
$$ \huge \begin{cases} x + y + z = 6 \\ 2x + 2y + 2z = 12 \\ x - y + z = 2 \end{cases} $$

Representação matricial usando ndarrays:
$$ \large
\begin{pmatrix} 1 & 1 & 1 \\ 2 & 2 & 2 \\ 1 & -1 & 1 \end{pmatrix}
\begin{pmatrix} x \\ y \\ z \end{pmatrix}
=
\begin{pmatrix} 6 \\ 12 \\ 2 \end{pmatrix}
\rightarrow
\begin{pmatrix} 1 & 1 & 1 & 6 \\ 2 & 2 & 2 & 12 \\ 1 & -1 & 1 & 2 \end{pmatrix}
$$
Setup e resolução:

````python
spi_sistema = np.array([[1, 1, 1, 6], [2, 2, 2, 12], [1, -1, 1, 2]], dtype=float)
print(f"==========Sistema original==========")
imprimir_sistema(spi_sistema)
spi_sistema_escalonado = escalonar_sistema(spi_sistema)
print(f"\n\n==========Sistema escalonado==========")
imprimir_sistema(spi_sistema_escalonado)
spi_resultado = resolver(spi_sistema_escalonado)
print(f"\n\nResultado:")
imprimir_resultado(spi_resultado)
````

Resultado final:
$$ \huge \begin{align} x + z &= 4 && y = 2 && z = \text{livre} \end{align} $$

## 3.3 Exemplo SI
Sistema original:
$$ \huge \begin{cases} x + y + z = 3 \\ 2x + 2y + 2z = 6 \\ x + y + z = 4 \end{cases} $$

Representação matricial usando ndarrays:
$$ \large
\begin{pmatrix} 1 & 1 & 1 \\ 2 & 2 & 2 \\ 1 & 1 & 1 \end{pmatrix}
\begin{pmatrix} x \\ y \\ z \end{pmatrix}
=
\begin{pmatrix} 3 \\ 6 \\ 4 \end{pmatrix}
\rightarrow
\begin{pmatrix} 1 & 1 & 1 & 3 \\ 2 & 2 & 2 & 6 \\ 1 & 1 & 1 & 4 \end{pmatrix}
$$

Setup e resolução:

````python
si_sistema = np.array([[1, 1, 1, 3], [2, 2, 2, 6], [1, 1, 1, 4]], dtype=int)
print(f"==========Sistema original==========")
imprimir_sistema(si_sistema)
si_sistema_escalonado = escalonar_sistema(si_sistema)
print(f"\n\n==========Sistema escalonado==========")
imprimir_sistema(si_sistema_escalonado)
si_resultado = resolver(si_sistema_escalonado)
print(f"\n\nResultado:")
imprimir_resultado(si_resultado)
````

Resultado do escalonamento:
$$ \huge \begin{align} x + y + z &= 3 && 0 = 0 && 0 = 1 \end{align} $$
